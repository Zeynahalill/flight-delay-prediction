# Flight Delay Prediction -- Workshop Walkthrough
Bu notebook, `src/flight_delay` modüllerini kullanarak pipeline'ı adım adım gösterir.

```python
from flight_delay import config, data_loading, data_validation, eda, training, evaluation, predict
```

## 1. Dataset

```python
df = data_loading.load_raw_data()
data_validation.validate_schema(df)
df.head()
```

## 2. EDA

```python
eda.run_eda(df)
```

## 3. Train / Validation / Test Split

```python
train_df, val_df, test_df = training.split_data(df)
len(train_df), len(val_df), len(test_df)
```

## 4. Model Eğitimi + Hyperparameter Tuning

```python
models, preprocessor, X_test, y_test = training.train_all(train_df, val_df, test_df)
```

## 5. Evaluation

```python
results = evaluation.evaluate_all(models, X_test, y_test, feature_names=config.FEATURE_COLUMNS)
results
```

## 6. Modeli Kaydet

```python
best_name = max(results, key=lambda k: results[k]["f1"])
training.save_artifact(models[best_name], preprocessor)
print("Kaydedilen model:", best_name)
```

## 7. Yeni Veri ile Tahmin (Canlı Demo)

```python
proba, label = predict.predict_new({
    "Airline": "AA", "AirportFrom": "JFK", "AirportTo": "LAX",
    "DayOfWeek": 5, "Time": 1140, "Length": 240,
})
print(f"Gecikme olasılığı: {proba:.2%} -> {label}")
```
